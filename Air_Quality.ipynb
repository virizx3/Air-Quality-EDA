#data cleaning

## Create Clean Dataset

```python
df_clean = df.copy()
```

## Remove Completely Missing Column

```python
df_clean = df_clean.drop(columns=["Message"])
```

## Convert Start Date

```python
df_clean["Start_Date"] = pd.to_datetime(
    df_clean["Start_Date"],
    errors="coerce"
)
```

## Standardize Measurement Units

```python
df_clean["Measure Info"] = df_clean["Measure Info"].replace(
    {"µg/m3": "mcg/m3"}
)
```

## Check Text Formatting

```python
text_cols = df_clean.select_dtypes(include="str").columns

for col in text_cols:
    print(
        col,
        (df_clean[col] != df_clean[col].str.strip()).sum()
    )
```

## Final Data Quality Check

```python
print("Shape:", df_clean.shape)

print("\nMissing values:")
print(df_clean.isnull().sum())

print("\nDuplicate rows:", df_clean.duplicated().sum())

print(
    "Duplicate Unique IDs:",
    df_clean["Unique ID"].duplicated().sum()
)
```

#FEATURE ENGINEERING

```python
df_clean["Start Year"] = df_clean["Start_Date"].dt.year
```

```python
df_clean[["Start_Date", "Time Period", "Start Year"]].head(10)
```

```python
df_clean["Period Type"] = df_clean["Time Period"].apply(
    lambda x:
        "Summer" if x.startswith("Summer")
        else "Winter" if x.startswith("Winter")
        else "Annual" if x.startswith("Annual")
        else "Year/Range"
)
```

```python
df_clean[["Time Period", "Period Type"]].drop_duplicates().head(20)
```

```python
df_clean["Is Multi-Year"] = df_clean["Time Period"].str.contains(
    r"\d{4}-\d{2,4}",
    regex=True
)
```

```python
df_clean[["Time Period", "Start Year", "Period Type", "Is Multi-Year"]].drop_duplicates().head(20)
```

#DATA ANALYSIS

##How do PM2.5 values vary across different geographic types?

```python
pm25 = df_clean[df_clean["Name"] == "Fine particles (PM 2.5)"]
```

```python
pm25.groupby("Geo Type Name")["Data Value"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)
```

```python
fig = px.box(
    pm25,
    x="Geo Type Name",
    y="Data Value",
    title="PM2.5 Values by Geographic Type"
)

fig.show()
```

##How do PM2.5 values change over time?

```python
pm25.groupby("Start Year")["Data Value"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)
```

```python
fig = px.line(
    pm25.groupby("Start Year", as_index=False)["Data Value"].mean(),
    x="Start Year",
    y="Data Value",
    markers=True,
    text="Data Value",
    title="Average PM2.5 Values Over Time"
)

fig.update_traces(
    texttemplate="%{text:.2f}",
    textposition="top center"
)

fig.update_layout(
    xaxis_title="Start Year",
    yaxis_title="Average PM2.5"
)

fig.show()
```

##How do PM2.5 values vary by period type?

```python
pm25.groupby("Period Type")["Data Value"].agg(
    ["count", "mean", "median", "min", "max"]
).round(2)
```

##Which geographic type shows the highest variability in PM2.5?

```python
pm25.groupby("Geo Type Name")["Data Value"].agg(
    ["count", "mean", "std", "min", "max"]
).round(2)
```

##How does PM2.5 vary across geographic types and period types?

```python
pm25.groupby(
    ["Geo Type Name", "Period Type"]
)["Data Value"].mean().round(2).unstack()
```

##Which PM2.5 geographic type has the highest average value over time?

```python
pm25_year_geo = pm25.groupby(
    ["Start Year", "Geo Type Name"],
    as_index=False
)["Data Value"].mean()

pm25_year_geo.head()
```

```python
fig = px.line(
    pm25_year_geo,
    x="Start Year",
    y="Data Value",
    color="Geo Type Name",
    markers=True,
    title="Average PM2.5 by Geographic Type Over Time"
)

fig.update_layout(
    xaxis_title="Start Year",
    yaxis_title="Average PM2.5"
)

fig.show()
```
